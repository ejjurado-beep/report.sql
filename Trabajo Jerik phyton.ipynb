## Section A: Functions with main()

### Why do we need a main() function?

The `main()` function is the **entry point** of your program. It:
- Organizes your code
- Makes it easy to read and understand
- Follows best practices used by professional programmers

**Basic structure:**
```python

    # Your code here
    print("Hello!")

main()  # Call the function
```

### Example 1: Simple Greeting Function

Let's create a helper function and call it from main():

````python
def greet(name):
    # This function takes a name and prints a greeting
    print(f"Hello, {name}! Welcome to Python! 👋")

def main():
    # Call the greet function from main
    greet("Alice")
    greet("Bob")

main()
````

## Section B: Return Types (Not Just Strings!)

### What does "return" mean?

The `return` statement sends a value back from a function. It can be:
- A **string** (text)
- An **integer** (whole number)
- A **boolean** (True/False)
- A **list** (collection of items)
- Or other types!

**Key idea:** Don't always return strings. Choose the data type that makes sense! 🎯

### Example 2: Different Return Types

````python
def add_numbers(a, b):
    # This function returns an INTEGER, not a string
    result = a + b
    return result

def is_even(number):
    # This function returns a BOOLEAN (True or False)
    return number % 2 == 0

def get_favorite_fruits():
    # This function returns a LIST
    return ["apple", "banana", "strawberry"]

def main():
    # Using the functions
    sum_result = add_numbers(5, 3)
    print(f"5 + 3 = {sum_result}")
    print(f"Type: {type(sum_result)}") # Shows it's an int
    
    even_check = is_even(10)
    print(f"\nIs 10 even? {even_check}")
    print(f"Type: {type(even_check)}") # Shows it's a bool
    
    fruits = get_favorite_fruits()
    print(f"\nFavorite fruits: {fruits}")
    print(f"Type: {type(fruits)}") # Shows it's a list

main()
````

## Section C: Lists (The Basics)

### Problem: Managing Too Many Variables

Imagine you're a teacher with 5 students and their grades. Without lists, you might write:

```python
grade_1 = 85
grade_2 = 90
grade_3 = 78
grade_4 = 92
grade_5 = 88
```

Now imagine you have **100 students**... 😱 That would be 100 variables! And if you want to calculate the average? Nightmare!

**Solution: Use a list!**

````python
# WITHOUT LISTS - The Problem
print("WITHOUT LISTS (5 students):")
print("="*40)

grade_1 = 85
grade_2 = 90
grade_3 = 78
grade_4 = 92
grade_5 = 88

# To calculate average, we have to do this:
average = (grade_1 + grade_2 + grade_3 + grade_4 + grade_5) / 5
print(f"Average: {average}")

# To find the highest grade:
highest = max(grade_1, grade_2, grade_3, grade_4, grade_5)
print(f"Highest: {highest}")

print("\nImagine doing this with 100 students!!! 🤯")
````

### The Solution: Use Lists! 🎉

A **list** stores multiple values in one container:

```python
grades = [85, 90, 78, 92, 88]
```

Now we can:
- Calculate average easily
- Find the highest grade easily
- **Use the same code for 100 students!**

````python
# WITH LISTS - The Solution
print("WITH LISTS (same 5 students):")
print("="*40)

grades = [85, 90, 78, 92, 88]

# Calculate average - much simpler!
average = sum(grades) / len(grades)
print(f"Average: {average}")

# Find the highest grade:
highest = max(grades)
print(f"Highest: {highest}")

# Find the lowest grade:
lowest = min(grades)
print(f"Lowest: {lowest}")

print("\n✨ Same code works for 100 students!")
````

### Creating Lists and Accessing Elements

#### Creating a List

```python
# List with numbers
numbers = [1, 2, 3, 4, 5]

# List with strings
fruits = ["apple", "banana", "cherry"]

# Empty list (we'll add items later)
my_list = []
```

#### Accessing Elements

**Remember: Indexing starts at 0!**
- First item: `my_list[0]`
- Second item: `my_list[1]`
- Last item: `my_list[-1]`

````python
def show_list_access(items):
    # This function demonstrates how to access list elements
    print(f"Full list: {items}")
    print(f"First element: {items[0]}")
    print(f"Second element: {items[1]}")
    print(f"List length: {len(items)}")
    print(f"Last element: {items[-1]}")

def main():
    colors = ["red", "blue", "green", "yellow"]
    show_list_access(colors)

main()
````

### How to Loop Through a List

#### Method 1: For-In Loop (Simplest)

```python
for item in my_list:
    print(item)
```
Use this when you just need the **value** of each item.

#### Method 2: For with Range (When You Need the Index)

```python
for i in range(len(my_list)):
    print(f"Position {i}: {my_list[i]}")
```
Use this when you need to know **which position** each item is at.

````python
def print_method_1(items):
    # Method 1: For-In Loop - Simple and clean
    print("Method 1: For-In Loop")
    for item in items:
        print(f"  {item}")

def print_method_2(items):
    # Method 2: For with Range (when you need the index)
    print("\nMethod 2: For with Range (and Index)")
    for i in range(len(items)):
        print(f"  Position {i}: {items[i]}")

def main():
    animals = ["dog", "cat", "bird", "fish"]
    print_method_1(animals)
    print_method_2(animals)

main()
````

### Simple List Applications: Sum, Average, Maximum

#### Example 1: Calculate the Sum

````python
def calculate_sum(numbers):
    # This function calculates the sum of all numbers in a list
    total = 0
    for num in numbers:
        total = total + num
    return total

def main():
    scores = [10, 20, 30, 40, 50]
    result = calculate_sum(scores)
    print(f"Scores: {scores}")
    print(f"Sum: {result}")
    print(f"\nAlternatively, Python has a built-in: {sum(scores)}")

main()
````

#### Example 2: Calculate the Average

````python
def calculate_average(numbers):
    # This function calculates the average of all numbers in a list
    # Average = Sum / Count
    if len(numbers) == 0:
        return 0  # Avoid division by zero
    total = sum(numbers)
    average = total / len(numbers)
    return average

def main():
    temperatures = [22, 24, 20, 25, 23]
    avg = calculate_average(temperatures)
    print(f"Temperatures: {temperatures}")
    print(f"Average: {avg}°C")

main()
````

#### Example 3: Find the Maximum (Highest Value)

````python
def find_maximum(numbers):
    # This function finds the highest value in a list
    if len(numbers) == 0:
        return None
    highest = numbers[0]  # Start with the first number
    for num in numbers:
        if num > highest:
            highest = num
    return highest

def main():
    sales = [150, 200, 175, 220, 195]
    max_sale = find_maximum(sales)
    print(f"Sales: {sales}")
    print(f"Highest sale: ${max_sale}")
    print(f"\nAlternatively, Python has a built-in: {max(sales)}")

main()
````

### Modifying Lists: Change, Add, and Remove

#### Change an Element

````python
# Change an element by using its index
fruits = ["apple", "banana", "cherry"]
print(f"Original: {fruits}")

fruits[1] = "orange"  # Change the second item
print(f"After change: {fruits}")
````

#### Add an Element with append()

````python
def add_item_to_list(items, new_item):
    # This function adds a new item to the end of a list
    items.append(new_item)
    return items

def main():
    colors = ["red", "blue", "green"]
    print(f"Before: {colors}")
    
    colors = add_item_to_list(colors, "yellow")
    print(f"After append: {colors}")
    
    colors = add_item_to_list(colors, "purple")
    print(f"After another append: {colors}")

main()
````

#### Remove an Element with pop()

The `pop()` method removes and returns the **last item** in a list.

```python
my_list.pop()  # Removes the last item
```

````python
def remove_last_item(items):
    # This function removes the last item from a list
    # It returns what was removed
    removed = items.pop()
    return removed

def main():
    tasks = ["study", "exercise", "eat", "sleep"]
    print(f"Tasks: {tasks}")
    
    last_task = remove_last_item(tasks)
    print(f"\nRemoved: {last_task}")
    print(f"Remaining: {tasks}")
    
    another_removed = remove_last_item(tasks)
    print(f"\nRemoved: {another_removed}")
    print(f"Remaining: {tasks}")

main()
````

## Section D: Strings as Lists

### Problem: Getting Letters from a Name

Imagine you want to personalize a greeting for each friend. You need to get specific letters from their names. How can you do that in Python?

### Concept: Strings ARE Sequences (Like Lists!)

A **string** in Python is a sequence of characters, just like a list. You can access each character using its **index** (position), just like with lists.

#### Access Characters by Index

````python
def show_characters(text):
    # Show how to access specific characters
    print(f"Text: {text}")
    print(f"First character: {text[0]}")
    print(f"Second character: {text[1]}")
    print(f"Last character: {text[-1]}")

def main():
    name = "Python"
    show_characters(name)

main()
````

#### Loop Through Each Character

````python
def print_each_letter(text):
    # Print each letter on a new line
    print(f"Letters in '{text}':")
    for letter in text:
        print(f"  {letter}")

def main():
    word = "hello"
    print_each_letter(word)

main()
````

#### Extract Part of a String (Slicing)

````python
def get_substring(text):
    # Get a portion of the string
    print(f"Full text: {text}")
    print(f"First 3 letters: {text[0:3]}")
    print(f"Last 2 letters: {text[-2:]}")

def main():
    greeting = "Welcome"
    get_substring(greeting)

main()
````

### Key Difference: Strings vs. Lists

Unlike lists, **strings cannot be changed** (they are immutable). If you try this, Python will give you an error:

```python
text = "hello"
# This will cause an error:
# text[0] = "H"
```

But you can **create a new string** if you need to make changes.

### Compare Lists and Strings

| Operation | Lists | Strings |
|-----------|-------|----------|
| Access by index | ✅ Yes | ✅ Yes |
| Loop through | ✅ Yes | ✅ Yes |
| Slicing | ✅ Yes | ✅ Yes |
| Change an element | ✅ Yes | ❌ No |


### Your Turn: Work with Strings

**Task:** Write a function called `count_vowels(text)` that:
1. Takes a string as input
2. Counts how many vowels (a, e, i, o, u) are in it
3. Returns the count as an INTEGER

Hint: Use a loop to check each character! 🔤

````python
# Write your code here!

def count_vowels(text):
    # Count and return the number of vowels
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count

def main():
    # Test with different strings
    sentences = [
        "Python is fun",
        "Hola Mundo",
        "Functions and lists",
        "AEIOU aeiou"
    ]

    for s in sentences:
        print(f"'{s}' has {count_vowels(s)} vowels")

main()
````

## 🎯 Integration Challenge: Putting It All Together!

Now let's combine everything: **functions, lists, and strings**.

### Challenge: Analyze a List of Words

Write a program that:
1. Has a function `longest_word(words)` that finds and returns the longest word in a list
2. Has a function `word_lengths(words)` that returns a **list** of integers (the length of each word)
3. Has a function `reverse_all_words(words)` that returns a **list** of all words reversed
4. Calls these functions from main() and displays the results

**Bonus:** Add comments explaining what each function does! 🎊

````python
# Write your integration challenge here!

def longest_word(words):
    # Find and return the longest word
    if not words:
        return None
    return max(words, key=len)

def word_lengths(words):
    # Return a list of the length of each word
    lengths = []
    for word in words:
        lengths.append(len(word))
    return lengths

def reverse_all_words(words):
    # Return a list of all words reversed
    reversed_words = []
    for word in words:
        reversed_words.append(word[::-1])
    return reversed_words

def main():
    # Create a list of words
    words = ["Python", "is", "awesome", "and", "powerful"]

    # Call your functions and display results
    print("Words:", words)
    print("Longest word:", longest_word(words))
    print("Word lengths:", word_lengths(words))
    print("Reversed words:", reverse_all_words(words))

main()
````
